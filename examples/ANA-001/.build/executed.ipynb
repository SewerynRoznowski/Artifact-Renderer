# ANA-001 — Payload power and duty-cycle allocation

Allocates peak power and duty cycle to each of the three payloads, establishing that the satellite bus provides sufficient electrical power to all payloads in all operational modes (`REQ-MIS-001`).

```python
import capella_cubed as cubed
%matplotlib inline

ana = cubed.load()
ana.context_diagram
```

```python
ana.assume(alias="num_monkeys", value=5, unit= "count", rationale="Since we need so many dabloons, we surely will need this many monkeys")
```

```python
ana.assumptions
```

```python
ana.plot_scores()
```

```python
dabloons = float(ana.value_of("REQ-SYS-001"))
num_monkeys = ana.value_of("num_monkeys")
```

# Now we calculate how many bomb-bananas we need

```python
bomb_bananas = num_monkeys * dabloons

ana.guarantee("REQ-SYS-002", bomb_bananas)
```
